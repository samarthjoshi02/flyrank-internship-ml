## 2. The question: decision, action, cost of a wrong call

- **Research Question:** Which visible pages should be reviewed first for a refresh or expansion based on recent performance declines and available search opportunity?
- **Decision being made:** Prioritizing which pages an editor should review and update.
- **Unit of Analysis:** A single pseudonymized content item (page).
- **Expected Model Output:** A ranked priority queue of pages (with reason codes) indicating which pages most urgently need a refresh.
- **Action someone would take:** An editor reviews the top K recommended pages in the queue and updates the content, meta description, or structure based on the reason code.
- **Cost of False Positive:** An editor wastes limited time reviewing and attempting to fix a page that was either already fine or declining due to seasonal noise.
- **Cost of False Negative:** A truly declining page that could have been saved is ignored, leading to continued traffic loss.
- **Why Machine Learning is useful instead of simple rules:** While a simple rule-based score gets roughly 12 out of 50 top recommendations correct (Precision@50 = 0.240), an ML ranking model can balance complex interactions between age, current volume, trend trajectory, and content properties simultaneously to achieve much higher precision (e.g., Random Forest yielding 0.740).

## 3. Quick look at the data (2-3 real numbers)

The dataset provides 30,000 pseudonymized pages across 32 clients. A significant portion of the dataset (54.2%) is labeled as 'down' in trend direction. The average trend percentage across the slice is roughly -4.78%, indicating a broad pattern of decline that makes this an opportunity-rich dataset for a refresh model. Missing data is systematic; for example, keyword search volume is missing for 2,468 rows, heavily correlated with certain content types.

```python
import pandas as pd

# Load the dataset
df = pd.read_csv('../../data/raw/content_refresh_anonymized.csv')

# Calculate basic statistics
num_rows, num_cols = df.shape
num_clients = df['client_id'].nunique()
percent_declining = (df['trend_direction'] == 'down').mean() * 100
avg_trend = df['trend_pct'].mean()
missing_search_volume = df['search_volume'].isnull().sum()

print("--- Quick Look at the Data ---")
print(f"Number of rows: {num_rows}")
print(f"Number of columns: {num_cols}")
print(f"Number of unique clients: {num_clients}")
print(f"Percentage of declining pages: {percent_declining:.1f}%")
print(f"Average trend percentage: {avg_trend:.2f}%")
print(f"Rows missing search volume: {missing_search_volume}")

# Distribution of labels
print("\n--- Distribution of Trend Direction ---")
print(df['trend_direction'].value_counts())
```

## 4. Careful words: what I can and can't claim

- **Correlation is not causation:** We observe that certain page properties (like low word count) are associated with declining traffic, but we have not run an experiment to prove that changing them guarantees recovery.
- **Model predictions assist decision-making:** The output is a decision-support tool designed to surface candidates for human review, not an automated magic SEO fixer.
- **Dataset limitations:** The dataset is an unbalanced panel across clients. Our label `is_declining_label` is a proxy based on recent trailing windows rather than a strict forward-looking future outcome.
- **Possible bias:** Missing data in our features (such as keyword properties) is systematic, not random. Blindly imputing missing values could inject artificial signals into the model.
- **Generalisation limitations:** Results obtained on this 30k-row starter dataset might not perfectly scale to the full 79M-row warehouse release. We are not claiming to have reverse-engineered or 'predicted' Google's algorithm.

```python
# We will ensure safe claims by keeping track of our data limitations.
safe_claims = [
    "Observed correlation between page properties and decline.",
    "Decision-support ranking for human reviewers.",
    "Directional indicators of opportunity."
]

unsafe_claims = [
    "Proof that a refresh causes recovery.",
    "Reverse-engineering Google's search algorithm.",
    "Guaranteed traffic increases."
]

print("Allowed Claims:")
for claim in safe_claims:
    print(f" ✓ {claim}")
    
print("\nDisallowed Claims:")
for claim in unsafe_claims:
    print(f" ✗ {claim}")
```
